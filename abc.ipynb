```python
import torch
import torch.nn.functional as F
```

```python
# torch.save(r2.detach().cpu(), "r2_100.pt")
```

```python
r2_100 = torch.load("r2_100.pt", map_location="cpu")
r2_none = torch.load("r2_none.pt", map_location="cpu")
```

```python
rmse = torch.sqrt(torch.mean((r2_100 - r2_none) ** 2)).item()

cosine = F.cosine_similarity(
    r2_100.flatten(),
    r2_none.flatten(),
    dim=0
).item()

print("\nRMSE:", rmse)
print("Cosine similarity:", cosine)

print("\nr2_100 abs mean:", r2_100.abs().mean().item())
print("r2_none abs mean:", r2_none.abs().mean().item())

print("r2_100 std:", r2_100.std().item())
print("r2_none std:", r2_none.std().item())

print("r2_100 max abs:", r2_100.abs().max().item())
print("r2_none max abs:", r2_none.abs().max().item())
```
